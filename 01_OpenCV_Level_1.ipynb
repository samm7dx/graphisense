```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

print("GraphiSense environment ready")
print("OpenCV:", cv2.__version__)
print("NumPy:", np.__version__)
```

```python
dataset_dir = Path("data/raw/handwriting/extracted")

print("Dataset exists:", dataset_dir.exists())
print("Dataset location:", dataset_dir.resolve())
```

```python
images = list(dataset_dir.rglob("*.jpg"))

print("Total images:", len(images))
```

```python
for i, path in enumerate(images[:5]):
    print(i, path)
```

```python
image_path = images[0]

image = cv2.imread(str(image_path))

print("Path:", image_path)
print("Loaded:", image is not None)
print("Shape:", image.shape)
print("Data type:", image.dtype)
```

```python
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(12, 6))
plt.imshow(image_rgb)
plt.axis("off")
plt.show()
```
